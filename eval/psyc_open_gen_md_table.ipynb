```python
import pandas as pd
import numpy as np

from llm_audit import BASE_DIR


DATA_PATH = BASE_DIR / "eval" / "data" / "boot" / "psyc_open_en.csv"


def fmt(mean, lo, hi, dec=2):
    f = f".{dec}f"
    return f"{mean:{f}} [{lo:{f}},{hi:{f}}]"


df = pd.read_csv(DATA_PATH)

metrics = ["1D", "AGR", "SUB", "CONV"]
header = ["Model"] + [col for m in metrics for col in (m, f"{m}_adj [CI]")]
sep = [":---"] + ["---:"] * (len(metrics) * 2)

lines = []
lines.append("|" + "|".join(header) + "|")
lines.append("|" + "|".join(sep) + "|")

for _, row in df.iterrows():
    cells = [row["Model"]]
    for m in metrics:
        cells.append(f"{row[m]:.2f}")
        cells.append(fmt(row[f"{m}_adj"], row[f"{m}_adj_ci_lo"], row[f"{m}_adj_ci_up"]))
    lines.append("|" + "|".join(cells) + "|")

table = "\n".join(lines)
print(table)
```
